```python
# TODO B2 — perbaiki satu kolom ordinal yang salah tipe
# Pada dataset genshin, kelompok_umur_rilis merupakan variabel ordinal
urutan_umur = ['<1 Tahun', '2-3 Tahun', '4+ Tahun']
df['kelompok_umur_rilis'] = pd.Categorical(
    df['kelompok_umur_rilis'],
    categories=urutan_umur,
    ordered=True
)

print(df['kelompok_umur_rilis'].value_counts().sort_index())
print()

median_code = df['kelompok_umur_rilis'].cat.codes.median()
median_category = urutan_umur[int(round(median_code))]
print("Median kode:", median_code)
print("Median kelompok umur rilis:", median_category)
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'ganti-dengan-kalimat-rahasia-tim'
```

```python
# TODO E2 — HASH
def samarkan(nilai, garam):
    if pd.isnull(nilai):
        return np.nan
    gabungan = str(garam) + str(nilai)
    hash_obj = hashlib.sha256(gabungan.encode('utf-8'))
    return hash_obj.hexdigest()[:12]

GARAM = 'rahasiatim_2025_genshin_eda'
aman['name'] = aman['name'].apply(lambda x: samarkan(x, GARAM))
display(aman[['name']].head())
```

```python
# TODO E2 — HASH
def samarkan(nilai, garam):
    if pd.isnull(nilai):
        return np.nan
    gabungan = str(garam) + str(nilai)
    hash_obj = hashlib.sha256(gabungan.encode('utf-8'))
    return hash_obj.hexdigest()[:12]

GARAM = 'rahasiatim_2025_eda_kmt'

# Terapkan hash ke kolom bertindakan 'hash' (misalnya id_peserta)
kolom_hash = [col for col, (kat, tind) in PENANDAAN.items() if tind == 'hash']
for col in kolom_hash:
    if col in aman.columns:
        aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

```

```python
# TODO E3 — buktikan hash bersifat konsisten
hash_1 = samarkan('Hero_A', GARAM)
hash_2 = samarkan('Hero_A', GARAM)
print(f"Konsisten? : {hash_1 == hash_2}")
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
hash_1 = samarkan('PLT-2025-1309', GARAM)
hash_2 = samarkan('PLT-2025-1309', GARAM)
print(f"Hasil hash pertama : {hash_1}")
print(f"Hasil hash kedua   : {hash_2}")
print(f"Konsisten/Identik? : {hash_1 == hash_2}")
print()

# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda
hash_3 = samarkan('PLT-2025-1512', GARAM)
print(f"Hasil hash nilai lain: {hash_3}")
print(f"Berbeda?             : {hash_1 != hash_3}")
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

```

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
def cek_k(data, kolom):
    grup = data.groupby(kolom).size()
    k_value = grup.min()
    grup_1_orang = (grup == 1).sum()
    total_grup = len(grup)
    return pd.Series({
        'k': k_value,
        'grup_1_orang': grup_1_orang,
        'total_grup': total_grup
    })
```
