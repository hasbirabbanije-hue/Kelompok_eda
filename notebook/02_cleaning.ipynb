```python
import pandas as pd
import numpy as np
import os

pd.set_option('display.max_columns', 35)
pd.set_option('display.width', 140)

FILE = 'genshin_impact_clean.csv'
if not os.path.exists(FILE):
    try:
        from google.colab import files
        files.upload()
    except ImportError:
        raise FileNotFoundError(f'Letakkan {FILE} di folder yang sama dengan notebook ini.')

df = pd.read_csv(FILE)
df.shape
```

```python
kosong_tim = df.isna().sum()
kosong_tim_df = kosong_tim[kosong_tim > 0].to_frame('jumlah_kosong').assign(
    persen=lambda d: (d.jumlah_kosong / len(df) * 100).round(1)
)
print('--- Nilai Kosong Sejati ---')
display(kosong_tim_df)
```

```python
CURIGA_TIM = ['n/a', 'na', '-', '', 'null', 'none', 'tidak diketahui', 'unknown', '?', '#n/a']
print('--- Placeholder Teks ---')
for kol in df.select_dtypes(include='object').columns:
    nilai = df[kol].astype(str).str.strip().str.lower()
    ketemu = nilai[nilai.isin(CURIGA_TIM)].value_counts()
    if len(ketemu):
        print(f'{kol}:')
        print(ketemu.to_string(), '\n')

print('--- Keanehan Rentang Angka (Min/Max) ---')
for kol in df.select_dtypes(include='number').columns:
    s = df[kol]
    print(f'{kol:25s} min={s.min():12,.1f}  max={s.max():14,.1f}  '
          f'nilai terkecil unik: {sorted(s.dropna().unique())[:3]}')
```

```python
def pola_kosong(data, kolom_kosong, kolom_pembanding):
    return (data.groupby(kolom_pembanding)[kolom_kosong]
                .apply(lambda s: round(s.isna().mean() * 100, 1))
                .sort_values(ascending=False)
                .to_frame('persen_kosong'))

if len(kosong_tim_df) > 0:
    for kol_kosong in kosong_tim_df.index:
        print(f'\n--- Pola kekosongan {kol_kosong} berdasarkan element ---')
        display(pola_kosong(df, kol_kosong, 'element'))
else:
    print('\nTidak ada nilai kosong sejati (NaN), analisis pola kekosongan (MCAR/MAR/MNAR) tidak diperlukan.')
```

```python
print('\n--- Pemeriksaan Duplikasi ---')
print('Baris yang seluruh kolomnya duplikat persis:', df.duplicated().sum())
if 'name' in df.columns:
    print('Baris dengan nama karakter berulang:', df['name'].duplicated().sum())
```

```python
print('\n--- Konsistensi Kolom Teks (Spasi & Kapitalisasi) ---')
for kol in df.select_dtypes(include='object').columns:
    s = df[kol].astype(str)
    mentah = s.nunique()
    rapi = s.str.strip().str.lower().nunique()
    if mentah != rapi:
        print(f'{kol:25s} {mentah:4d} -> {rapi:4d} (selisih {mentah - rapi})')

for kol in ['weapon', 'element', 'region', '3D-model', 'asc_stat_bonus_category']:
    if kol in df.columns:
        print(f'\n--- Nilai Unik Kolom: {kol} ---')
        print(df[kol].value_counts(dropna=False))
```

```python
print('\n--- Pemeriksaan Tipe Data Tanggal ---')
if 'release_date' in df.columns:
    tgl_rilis = pd.to_datetime(df['release_date'], errors='coerce')
    print('Gagal dikonversi (NaN):', tgl_rilis.isna().sum())
    print('Rentang Tanggal Rilis:', tgl_rilis.min(), 'sampai', tgl_rilis.max())
    print('Bulan rilis unik      :', sorted(tgl_rilis.dt.month.dropna().unique()))
```

```python
print('\n--- Aturan Validitas Rentang ---')

aturan_tim = {}

if 'rarity' in df.columns:
    aturan_tim['Rarity harus 4 atau 5'] = ~df['rarity'].isin([4, 5])

if 'lvl_90_HP' in df.columns:
    aturan_tim['HP Level 90 harus > 0'] = df['lvl_90_HP'] <= 0
elif 'lvl_90_hp' in df.columns:
    aturan_tim['HP Level 90 harus > 0'] = df['lvl_90_hp'] <= 0

if 'lvl_90_ATK' in df.columns:
    aturan_tim['ATK Level 90 harus > 0'] = df['lvl_90_ATK'] <= 0
elif 'lvl_90_atk' in df.columns:
    aturan_tim['ATK Level 90 harus > 0'] = df['lvl_90_atk'] <= 0

if 'lvl_90_DEF' in df.columns:
    aturan_tim['DEF Level 90 harus > 0'] = df['lvl_90_DEF'] <= 0
elif 'lvl_90_def' in df.columns:
    aturan_tim['DEF Level 90 harus > 0'] = df['lvl_90_def'] <= 0

if 'duplicate_rate' in df.columns:
    aturan_tim['Duplicate rate harus 0-1'] = (df['duplicate_rate'] < 0) | (df['duplicate_rate'] > 1)

if 'c6_rate' in df.columns:
    aturan_tim['C6 rate harus 0-1'] = (df['c6_rate'] < 0) | (df['c6_rate'] > 1)

if 'num_banners' in df.columns:
    aturan_tim['Num banners harus >= 1'] = df['num_banners'] < 1

for nama, pelanggaran in aturan_tim.items():
    print(f'{nama:30s} dilanggar {int(pelanggaran.sum()):4d} baris')
```

```python
print('\n--- Aturan Hubungan Antarkolom ---')
kolom_c = [f'constellation_{i}_pull' for i in range(8)]
total_C = df[kolom_c].sum(axis=1)
tidak_sama = df['pulled_count'] != total_C
print('Baris yang jumlah pulled_count TIDAK sesuai dengan total C0-C7:', tidak_sama.sum())
if tidak_sama.sum() > 0:
    display(df.loc[tidak_sama, ['name', 'pulled_count'] + kolom_c])
```

```python
# 10.3 — Susun profil kualitas data tim dan simpan sebagai genshin_impact_kualitas_data.csv
temuan_tim = []

def catat_tim(dimensi, kolom, masalah, jumlah, keparahan, rencana):
    temuan_tim.append({
        'dimensi'   : dimensi,
        'kolom'     : kolom,
        'masalah'   : masalah,
        'jumlah'    : int(jumlah),
        'persen'    : round(jumlah / len(df) * 100, 2),
        'keparahan' : keparahan,
        'rencana'   : rencana,
    })

if len(kolom_c) == 8 and 'pulled_count' in df.columns:
    catat_tim('Akurasi', 'pulled_count', 'Jumlah total pull tidak sama dengan total C0-C7',
              tidak_sama.sum(), 'Tinggi',
              'Verifikasi sumber data rumus perhitungan pull atau hitung ulang total pull')

if 'release_date' in df.columns and df['release_date'].dtype == 'object':
    catat_tim('Validitas', 'release_date', 'Tipe data masih object/teks',
              len(df), 'Sedang',
              'Konversi kolom ke format datetime pada pertemuan 5')

region_aneh = df['region'].isin(['Nod-Krai', 'Extra-Terresterial']).sum() if 'region' in df.columns else 0
if region_aneh > 0:
    catat_tim('Konsistensi', 'region', 'Kategori region tidak standar (Nod-Krai, Extra-Terresterial)',
              region_aneh, 'Rendah',
              'Diskusikan dengan tim apakah kategori ini valid atau perlu dikelompokkan ke wilayah lain')

if 'Unnamed: 0' in df.columns:
    catat_tim('Keunikan', 'Unnamed: 0', 'Kolom indeks redundan terbawa dari export CSV',
              len(df), 'Rendah',
              'Hapus kolom Unnamed: 0 pada tahap pembersihan data')

catat_tim('Kelengkapan', 'Semua kolom', 'Pemeriksaan missing values sejati (NaN)',
          kosong_tim.sum(), 'Rendah',
          'Pertahankan struktur data jika 0, atau tangani sesuai mekanisme kekosongan')

profil_tim = pd.DataFrame(temuan_tim).sort_values(['keparahan', 'jumlah'], ascending=[False, False])
profil_tim.to_csv('genshin_impact_kualitas_data.csv', index=False)
print('\nTersimpan di genshin_impact_kualitas_data.csv dengan ukuran:', profil_tim.shape)
display(profil_tim)
```
